```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.metrics import r2_score

# 读取 CSV 文件
df = pd.read_csv(r"E:/Desktop/all_data/crosstalk_new/s5_ER_DHB_2nd.csv")

# 提取数据
x = df["inducer"]
y = df["RPU"]
LBD_values = df["LBD"].unique()  # 获取 LBD 列的唯一值

# 参数初始化
I = np.arange(0, 100, 0.001)
k1 = 6.353E-12


k2 = 3.583382137

k3= 0.002820379874017
kx1= 90.438632354216338
kx2= 0.000010055243956
I0 = 0.035485714

kd = 40.97704697


Imax = 35.84931505


# 存放所有预测值和真实值
all_y_true = []
all_y_pred = []

# 创建绘图窗口
fig, ax = plt.subplots(figsize=(5.08, 5.08))

# 为每个 LBD 分类绘制曲线和点
for L in LBD_values:
    # 筛选属于当前 LBD 分类的点
    subset = df[df["LBD"] == L]
    x_subset = subset["inducer"].values
    y_subset = subset["RPU"].values
   
    term_a = k2 * x_subset + kx1 + kx2 * k2 * x_subset + 1
    term_b = k2**2 * k3 * x_subset**2 + k1 + k1 * kx1**2 + k3 * k2**2 * kx2**2 * x_subset**2

    
    x1_pred = np.sqrt(term_a**2 + 8 * L * term_b)

    # 计算分子与分母
    ratio = (x1_pred - term_a) / (x1_pred + term_a)
    fraction = (L / 2) * (ratio * kd) / (1 + (L / 2) * (ratio * kd))

    # 预测表达量
    y_pred = Imax * fraction + I0


    all_y_true.extend(y_subset)
    all_y_pred.extend(y_pred)

    # 绘制曲线（用于展示）
    # 定义中间项，简化表达式
    term_a = k2*I+kx1+kx2*k2*I+1
    term_b = k2*k2*k3*I*I + k1+k1*kx1*kx1+k3*k2*k2*kx2*kx2*I*I
    # 计算中间变量 x1_curve
    x1_curve = np.sqrt(term_a**2 + 8 * L * term_b)

    # 计算比率项
    ratio = (x1_curve - term_a) / (x1_curve + term_a)

    # 计算主分数项（激活比例）
    fraction = (L / 2) * (ratio * kd) / (1 + (L / 2) * ratio * kd)

    # 预测表达量
    P1 = Imax * fraction + I0


    ax.plot(I, P1, label=f"LBD = {L:.2f}", linewidth=1)
    ax.scatter(x_subset, y_subset, s=15)

# 计算整体 R²
r2_total = r2_score(all_y_true, all_y_pred)

# 在图上显示 R²
ax.text(0.95, 0.05, f"$R^2 = {r2_total:.4f}$", transform=ax.transAxes,
        fontsize=12, verticalalignment='bottom', horizontalalignment='right',
        bbox=dict(boxstyle='round,pad=0.2', facecolor='white', edgecolor='gray'))

# 添加两条水平线
#ax.axhline(y=0.363935118, color='black', linestyle='--', linewidth=1)
#ax.axhline(y=2.120435836, color='black', linestyle='--', linewidth=1)

# 设置横纵坐标轴的取值范围和标签
ax.set_xscale("log")
ax.set_yscale("log")
ax.set_xlabel("Inducer", fontsize=10)
ax.set_ylabel("P*", fontsize=10)
ax.set_ylim(10**(-3), 10**2)

# 美化图形
ax.spines['bottom'].set_linewidth(0.5)
ax.spines['left'].set_linewidth(0.5)
ax.spines['top'].set_linewidth(0.5)
ax.spines['right'].set_linewidth(0.5)
plt.tick_params(axis='both', width=0.5, labelsize=10)
plt.title("s5-ER_DHB", fontsize=15)
plt.legend(fontsize=10)
plt.tight_layout()
plt.savefig('E:/Desktop/all_data/crosstalk_new/s5-ER_DHB_2nd.svg')
# 显示图形
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import os

# 固定参数
Imax = 35.84931505
kd = 40.97704697
I0 = 0.035485714  # 通用 I0

# Receptor 参数
receptors = ["ER", "DHBR", "PR", "GR", "MR"]
k1s =    [6.353e-12,    4.21691e-22,   1.22917e-05,  4.01e-06,     5.49128e-06]
k2s =    [3.583382137,  0.039590602,   0.462590457,  1.131061083,  2.00561142]
k3s =    [4.130148252, 13.17104327,    0.074821866,  4.16e-05,     0.001243206]
kx1s =   [1.112792988,  0.001007507,   0.00273789,   68.91003389,  1.937182665]
kx2s =   [4.960066803,  0.001604871,   0.921262549,  1.173481081,  1.619613647]
I_ranges = [
    (1e-4, 1e0),  # ER
    (1e-3, 1e2),  # DHBR
    (1e-3, 1e2),  # PR
    (1e-3, 1e2),  # GR
    (1e-6, 1e1),  # MR
]

# 统一 L 范围（所有图都用相同的横轴范围）
L_min, L_max = 1e-3, 1e2
L = np.logspace(np.log10(L_min), np.log10(L_max), 1000)

# 输出文件夹
output_dir = "E:/Desktop/LBD/+-inducer"
os.makedirs(output_dir, exist_ok=True)

# 绘图
for receptor, k1, k2, k3, kx1, kx2, (_, I_plus) in zip(receptors, k1s, k2s, k3s, kx1s, kx2s, I_ranges):

    P_list = []
    for I_val, label, color in zip([0, I_plus], ['-inducer', '+inducer'], ['gray', 'purple']):
        # 使用 I_val 表示当前诱导剂浓度（0 或 I_plus）
    
        x1 = np.sqrt((k2*I_val+kx1+kx2*k2*I_val+1)*(k2*I_val+kx1+kx2*k2*I_val+1)+ 8 * L * (k2*k2*k3*I_val*I_val + k1+k1*kx1*kx1+k3*k2*k2*kx2*kx2*I_val*I_val))


        P=(Imax * ((L / 2) * ((x1 -  (k2*I_val+kx1+kx2*k2*I_val+1)) / (x1 +  (k2*I_val+kx1+kx2*k2*I_val+1))) * kd / (1 + (L / 2) * ((x1 -  (k2*I_val+kx1+kx2*k2*I_val+1)) / (x1 +  (k2*I_val+kx1+kx2*k2*I_val+1))) *kd)) + I0)

    
        P_list.append((P, label, color))

    # 绘图
    fig, ax = plt.subplots(figsize=(5.08, 5.08))
    for P, label, color in P_list:
        ax.plot(L, P, linewidth=2, label=label, color=color)

    ax.set_xscale("log")
    ax.set_yscale("log")
    ax.set_xlabel("Ligand concentration (L)", fontsize=10)
    ax.set_ylabel("P*", fontsize=10)
    ax.set_ylim(1e-3, 1e2)
    ax.set_xlim(L_min, L_max)
    ax.set_title(f"{receptor}", fontsize=12)
    ax.legend()
    plt.tight_layout()

    # 保存图像
    fig.savefig(f"{output_dir}/{receptor}.svg", format="svg")
    plt.close(fig)

print("所有核受体图像（±inducer）已保存至：", output_dir)
```

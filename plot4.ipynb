```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.metrics import r2_score

# 读取 CSV 文件
df = pd.read_csv(r"E:\Desktop\\mamm\\410.csv")

# 提取数据
x = df["inducer"]
y = df["RPU"]
LBD_values = df["LBD"].unique()  # 获取 LBD 列的唯一值

# 参数初始化
I = np.arange(0, 2000, 0.01)
'''
k1= 0.000000024465372
k2= 11.007266886882983
k3= 25.868286102327847
kx1= 41.271599110563116
kx2= 12.813951781577774
'''
k1 = 3.48197E-05
k2 = 0.003561
k3 = 0.025226
Imax = 13.60910816
I0 = 0.01496709
kd = 10.56826129

# 存储所有预测值的容器
all_y_pred = np.array([])
all_y_true = np.array([])

# 创建绘图窗口
fig, ax = plt.subplots(figsize=(5.08, 5.08))

# 为每个 LBD 分类绘制曲线和点
for L in LBD_values:
    # 筛选属于当前 LBD 分类的点
    subset = df[df["LBD"] == L]
    x_subset = subset["inducer"].values
    y_subset = subset["RPU"].values
    
    # 计算 P1 曲线（用于绘图）
    x1 = np.sqrt((k2 * I + 1) ** 2 + 8 * L * (k2 * k2 * k3 * I * I + k1))
    P1 = (
        Imax
        * ((L / 2) * ((x1 - (k2 * I + 1)) / (x1 + (k2 * I + 1))) * kd
           / (1 + (L / 2) * ((x1 - (k2 * I + 1)) / (x1 + (k2 * I + 1))) * kd))
        + I0
    )
    
    # 绘制曲线
    ax.plot(I, P1, label=f"LBD = {L:.2f}", linewidth=2)
    
    # 绘制点
    ax.scatter(x_subset, y_subset, s=50)
    
    # 为每个数据点计算预测值（用于R²计算）
    for inducer_val, true_val in zip(x_subset, y_subset):
        x1_point = np.sqrt((k2 * inducer_val + 1) ** 2 + 8 * L * (k2 * k2 * k3 * inducer_val * inducer_val + k1))
        P1_point = (
            Imax
            * ((L / 2) * ((x1_point - (k2 * inducer_val + 1)) / (x1_point + (k2 * inducer_val + 1))) * kd
               / (1 + (L / 2) * ((x1_point - (k2 * inducer_val + 1)) / (x1_point + (k2 * inducer_val + 1))) * kd))
            + I0
        )
        all_y_pred = np.append(all_y_pred, P1_point)
        all_y_true = np.append(all_y_true, true_val)

# 添加两条水平线
ax.axhline(y=0.380931459, color='black', linestyle='--', linewidth=1)
ax.axhline(y=2.368348307, color='black', linestyle='--', linewidth=1)

# 计算线性尺度下的R²
r2_linear = r2_score(all_y_true, all_y_pred)

# 计算log10尺度下的R²
# 注意：需要确保所有值都为正数才能取log10
if np.all(all_y_true > 0) and np.all(all_y_pred > 0):
    log10_y_true = np.log10(all_y_true)
    log10_y_pred = np.log10(all_y_pred)
    r2_log10 = r2_score(log10_y_true, log10_y_pred)
else:
    r2_log10 = None
    print("警告：部分值小于等于0，无法计算log10 R²")

# 添加R²信息到图表中
if r2_log10 is not None:
    textstr = f'R² (linear) = {r2_linear:.4f}\nR² (log10) = {r2_log10:.4f}'
    print(f"线性R²值: {r2_linear:.4f}")
    print(f"log10 R²值: {r2_log10:.4f}")
else:
    textstr = f'R² (linear) = {r2_linear:.4f}\nR² (log10) = N/A'
    print(f"线性R²值: {r2_linear:.4f}")

props = dict(boxstyle='round', facecolor='wheat', alpha=0.8)
ax.text(0.05, 0.95, textstr, transform=ax.transAxes, fontsize=10,
        verticalalignment='top', bbox=props)

# 设置横纵坐标轴的取值范围和标签
ax.set_xscale("log")
ax.set_yscale("log")
ax.set_xlabel("Inducer", fontsize=12, weight='bold')
ax.set_ylabel("P*", fontsize=15, weight='bold')
ax.set_ylim(10**(-3), 10**2)

# 美化图形
ax.spines['bottom'].set_linewidth(2)
ax.spines['left'].set_linewidth(2)
ax.spines['top'].set_linewidth(2)
ax.spines['right'].set_linewidth(2)
plt.tick_params(axis='both', width=2, labelsize=10)
plt.title("s14-TraR", fontsize=15, weight='bold')

plt.legend(fontsize=10)
plt.tight_layout()

plt.savefig('E:\Desktop\mamm\curve\s14-TraR.svg')
# 显示图形
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.metrics import r2_score

# 读取 CSV 文件
df = pd.read_csv(r"E:\Desktop\mamm\414.csv")

# 提取数据
x = df["inducer"]
y = df["RPU"]
LBD_values = df["LBD"].unique()

# 参数初始化
k1 = 0.000484948812975
k2 = 58.409860
k3 = 0.048409
kx1 = 1.075915
kx2 = 5.632310
I0 = 0.064825979
kd = 9.152591138
Imax = 13.60910816

# 计算调整R²的函数
def adjusted_r2_score(y_true, y_pred, n_params):
    n = len(y_true)
    r2 = r2_score(y_true, y_pred)
    adj_r2 = 1 - (1 - r2) * (n - 1) / (n - n_params - 1)
    return r2, adj_r2

# 存放所有预测值和真实值
all_y_true = []
all_y_pred = []

# 创建绘图窗口
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10.16, 5.08))

# 为每个 LBD 分类绘制曲线和点
I = np.arange(0, 1, 0.00001)
for L in LBD_values:
    # 筛选属于当前 LBD 分类的点
    subset = df[df["LBD"] == L]
    x_subset = subset["inducer"].values
    y_subset = subset["RPU"].values
   
    term_a = k2 * x_subset + kx1 + kx2 * k2 * x_subset + 1
    term_b = k2**2 * k3 * x_subset**2 + k1 + k1 * kx1**2 + k3 * k2**2 * kx2**2 * x_subset**2
    x1_pred = np.sqrt(term_a**2 + 8 * L * term_b)
    ratio = (x1_pred - term_a) / (x1_pred + term_a)
    fraction = (L / 2) * (ratio * kd) / (1 + (L / 2) * (ratio * kd))
    y_pred = Imax * fraction + I0

    all_y_true.extend(y_subset)
    all_y_pred.extend(y_pred)

    # 绘制理论曲线
    term_a_curve = k2*I+kx1+kx2*k2*I+1
    term_b_curve = k2*k2*k3*I*I + k1+k1*kx1*kx1+k3*k2*k2*kx2*kx2*I*I
    x1_curve = np.sqrt(term_a_curve**2 + 8 * L * term_b_curve)
    ratio_curve = (x1_curve - term_a_curve) / (x1_curve + term_a_curve)
    fraction_curve = (L / 2) * (ratio_curve * kd) / (1 + (L / 2) * ratio_curve * kd)
    P1 = Imax * fraction_curve + I0

    ax1.plot(I, P1, label=f"LBD = {L:.2f}", linewidth=2)
    ax1.scatter(x_subset, y_subset, s=50)

# 计算普通R²和调整R²
n_params = 8
r2_total, adj_r2_total = adjusted_r2_score(all_y_true, all_y_pred, n_params)

# === 校正 R²（线性拟合后）===
slope, intercept = np.polyfit(all_y_pred, all_y_true, deg=1)
y_pred_corrected = slope * np.array(all_y_pred) + intercept
r2_corrected = r2_score(all_y_true, y_pred_corrected)

# 在第一个图上显示 R²、调整R²和校正R²
ax1.text(0.95, 0.15, f"$R^2 = {r2_total:.4f}$", transform=ax1.transAxes,
        fontsize=12, verticalalignment='bottom', horizontalalignment='right',
        bbox=dict(boxstyle='round,pad=0.2', facecolor='white', edgecolor='gray'))

ax1.text(0.95, 0.08, f"Adj. $R^2 = {adj_r2_total:.4f}$", transform=ax1.transAxes,
        fontsize=12, verticalalignment='bottom', horizontalalignment='right',
        bbox=dict(boxstyle='round,pad=0.2', facecolor='white', edgecolor='gray'))

ax1.text(0.95, 0.01, f"Corr. $R^2 = {r2_corrected:.4f}$", transform=ax1.transAxes,
        fontsize=12, verticalalignment='bottom', horizontalalignment='right',
        bbox=dict(boxstyle='round,pad=0.2', facecolor='white', edgecolor='gray'))

# 设置第一个图的横纵坐标轴
ax1.set_xscale("log")
ax1.set_yscale("log")
ax1.set_xlabel("Inducer", fontsize=10)
ax1.set_ylabel("P*", fontsize=10)
ax1.set_ylim(10**(-3), 10**2)
ax1.set_title("Model Fit", fontsize=15)
ax1.legend(fontsize=8)

# 创建第二个图：预测值 vs 真实值
ax2.scatter(all_y_pred, all_y_true, alpha=0.7, label='Data points')

# 绘制理想线（y=x）
min_val = min(min(all_y_true), min(all_y_pred))
max_val = max(max(all_y_true), max(all_y_pred))
ax2.plot([min_val, max_val], [min_val, max_val], 'k--', alpha=0.5, label='Ideal (y=x)')

# 绘制校正线
x_fit = np.linspace(min_val, max_val, 100)
y_fit = slope * x_fit + intercept
ax2.plot(x_fit, y_fit, 'r-', label=f'Correction line (slope={slope:.4f})')

ax2.set_xlabel('Predicted Values', fontsize=10)
ax2.set_ylabel('True Values', fontsize=10)
ax2.set_title('Predicted vs True Values', fontsize=15)
ax2.legend()

# 美化图形
for ax in [ax1, ax2]:
    ax.spines['bottom'].set_linewidth(0.5)
    ax.spines['left'].set_linewidth(0.5)
    ax.spines['top'].set_linewidth(0.5)
    ax.spines['right'].set_linewidth(0.5)
    ax.tick_params(axis='both', width=0.5, labelsize=10)

plt.tight_layout()

# 在控制台输出详细统计信息和诊断
print("=" * 60)
print("模型拟合诊断报告")
print("=" * 60)
print(f"样本数量 (n): {len(all_y_true)}")
print(f"模型参数数量 (k): {n_params}")
print(f"普通 R²: {r2_total:.6f}")
print(f"调整 R²: {adj_r2_total:.6f}")
print(f"校正 R²: {r2_corrected:.6f}")
print(f"线性校正斜率: {slope:.6f}")
print(f"线性校正截距: {intercept:.6f}")

# 诊断信息
print("\n" + "=" * 60)
print("诊断分析:")
print("=" * 60)

if r2_total < 0:
    print("❌ 普通R²为负: 模型预测比简单使用平均值更差")
    
if abs(slope - 1) > 0.5:
    print(f"❌ 校正斜率 ({slope:.4f}) 远离1: 模型存在系统性比例偏差")
    
if abs(intercept) > 0.1 * np.mean(all_y_true):
    print(f"❌ 校正截距 ({intercept:.4f}) 较大: 模型存在系统性偏移")

if r2_corrected > 0.8:
    print("✅ 校正R²较高: 模型有能力捕捉数据趋势")
    print("   建议: 重新优化参数，特别是与比例和偏移相关的参数")

# 计算均方根误差 (RMSE) 和平均绝对误差 (MAE)
rmse = np.sqrt(np.mean((np.array(all_y_true) - np.array(all_y_pred)) ** 2))
mae = np.mean(np.abs(np.array(all_y_true) - np.array(all_y_pred)))
print(f"\n误差指标:")
print(f"均方根误差 (RMSE): {rmse:.6f}")
print(f"平均绝对误差 (MAE): {mae:.6f}")

# 显示图形
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

# 读取 CSV 文件
df = pd.read_csv(r"E:\Desktop\work\quantification&crossvalidation4\NR\\330_new.csv")

# 提取数据
LBD_values = df["LBD"].unique()  # 获取 LBD 列的唯一值

# 参数初始化
I = np.arange(0, 1, 0.00001)
k1= 0.0000149644104
kx1= 0.839373346333649

k2= 4.324739056451484
k3= 2.224565229325882
kx2= 6.346936204080153
Imax = 35.84931505
I0 = 0.035485714
kd = 40.97704697

# 创建绘图窗口
fig, ax = plt.subplots(figsize=(5.08, 5.08))

# 为每个 LBD 分类绘制曲线和点
for L in LBD_values:
    # 筛选属于当前 LBD 分类的点
    subset = df[df["LBD"] == L]
    x_subset = subset["inducer"].values
    y_subset = subset["RPU"].values
    
    # 计算曲线
    x1 = np.sqrt((k2 * I + kx1 + kx2 * k2 * I + 1) ** 2 +
                 8 * L * (k2 * k2 * k3 * I * I + k1 + k1 * kx1 * kx1 +
                          k3 * k2 * k2 * kx2 * kx2 * I * I))
    P1 = (
        Imax
        * ((L / 2) * ((x1 - (k2 * I + kx1 + kx2 * k2 * I + 1)) /
                      (x1 + (k2 * I + kx1 + kx2 * k2 * I + 1))) * kd
           / (1 + (L / 2) * ((x1 - (k2 * I + kx1 + kx2 * k2 * I + 1)) /
                             (x1 + (k2 * I + kx1 + kx2 * k2 * I + 1))) * kd))
        + I0
    )
    
    # 根据实验 inducer 计算预测值
    x1_exp = np.sqrt((k2 * x_subset + kx1 + kx2 * k2 * x_subset + 1) ** 2 +
                     8 * L * (k2 * k2 * k3 * x_subset * x_subset + k1 + k1 * kx1 * kx1 +
                              k3 * k2 * k2 * kx2 * kx2 * x_subset * x_subset))
    y_pred = (
        Imax
        * ((L / 2) * ((x1_exp - (k2 * x_subset + kx1 + kx2 * k2 * x_subset + 1)) /
                      (x1_exp + (k2 * x_subset + kx1 + kx2 * k2 * x_subset + 1))) * kd
           / (1 + (L / 2) * ((x1_exp - (k2 * x_subset + kx1 + kx2 * k2 * x_subset + 1)) /
                             (x1_exp + (k2 * x_subset + kx1 + kx2 * k2 * x_subset + 1))) * kd))
        + I0
    )
    
    # 计算 R²
    ss_res = np.sum((y_subset - y_pred) ** 2)
    ss_tot = np.sum((y_subset - np.mean(y_subset)) ** 2)
    r2 = 1 - ss_res / ss_tot if ss_tot != 0 else np.nan
    
    # 绘制曲线
    ax.plot(I, P1, label=f"LBD={L:.4f}, R²={r2:.3f}", linewidth=2)
    # 绘制点
    ax.scatter(x_subset, y_subset, s=50)

# 设置坐标轴
ax.set_xscale("log")
ax.set_yscale("log")
ax.set_xlabel("inducer", fontsize=10)
ax.set_ylabel("P*", fontsize=10)
ax.set_ylim(10**(-3), 10**2)
ax.set_title("ER_s5", fontsize=15)

plt.legend()
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.metrics import r2_score
import matplotlib.cm as cm
import os

# 读取 CSV 文件
df = pd.read_csv(r"E:\\Desktop\\all_data\\crossvalidation\\374.csv")

# 提取数据
x = df["inducer"]
y = df["RPU"]
LBD_values = np.sort(df["LBD"].unique())

# 参数初始化
I = np.arange(0, 10, 0.001)
k1 = 0.000171923



k2 = 0.256016091


k3 = 225.0817618
kd = 1.445352554


Imax = 35.84931505
I0 = 0.111188222


# 创建绘图窗口
fig, ax = plt.subplots(figsize=(5.08, 5.08))

P1_all = []
y_all = []

# 使用颜色映射生成颜色从浅到深的渐变
colors = cm.Blues(np.linspace(0.3, 1, len(LBD_values)))

# 为每个 LBD 分类绘制曲线和点
for i, L in enumerate(LBD_values):
    subset = df[df["LBD"] == L]
    x_subset = subset["inducer"]
    y_subset = subset["RPU"]
    
    # 计算 P1 曲线
    x1 = np.sqrt((k2 * I + 1) ** 2 + 8 * L * (k2 * k2 * k3 * I * I + k1))
    P1 = (
        Imax
        * ((L / 2) * ((x1 - (k2 * I + 1)) / (x1 + (k2 * I + 1))) * kd
           / (1 + (L / 2) * ((x1 - (k2 * I + 1)) / (x1 + (k2 * I + 1))) * kd))
        + I0
    )
    P1_all.extend(np.interp(x_subset, I, P1))
    y_all.extend(y_subset)

    # 绘制曲线和点，颜色由浅到深
    ax.plot(I, P1, color=colors[i], label=f"LBD = {L:.2f}", linewidth=2)
    ax.scatter(x_subset, y_subset, color=colors[i], s=20)

# 计算总 R^2 值
r2 = r2_score(y_all, P1_all)

# 设置横纵坐标轴的取值范围和标签
ax.set_xscale("log")
ax.set_yscale("log")
ax.set_xlabel("Inducer", fontsize=10)
ax.set_ylabel("P*", fontsize=10)
ax.set_ylim(10**(-2), 10**2)

# 美化图形
ax.spines['bottom'].set_linewidth(0.5)
ax.spines['left'].set_linewidth(0.5)
ax.spines['top'].set_linewidth(0.5)
ax.spines['right'].set_linewidth(0.5)
plt.tick_params(axis='both', width=0.5, labelsize=10)
plot_title = f"LasR_s23 (R² = {r2:.4f})"
plt.title(plot_title, fontsize=12)
#plt.legend(fontsize=8)
plt.tight_layout()

# 保存图形到桌面固定文件夹，使用标题名命名文件名
desktop_path = r"C:\\Users\\Lenovo\\Desktop\\all_sensor_SI"
os.makedirs(desktop_path, exist_ok=True)
file_name = f"{plot_title.replace(' ', '_').replace('(', '').replace(')', '').replace('=', '').replace('²', '2')}.svg"
output_path = os.path.join(desktop_path, file_name)
plt.savefig(output_path, format='svg')

# 显示图形
plt.show()
```

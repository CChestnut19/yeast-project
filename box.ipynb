```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.colors as colors

def rob_phillips_allosteric_dimer_fixed_inducer(R, DBD, N_NS, delta_epsilon_rd, delta_epsilon_int, 
                                              K, P, delta_epsilon_pd, Imax, beta=1.0/0.6):
    """
    基于Rob Phillips的变构二聚体模型计算表达水平，诱导剂浓度固定
    
    参数:
    R: 转录因子浓度
    DBD: DBD浓度
    N_NS: 非特异性结合位点数量
    delta_epsilon_rd: TF-DNA结合自由能变化
    delta_epsilon_int: 二聚体相互作用自由能变化
    K: 变构常数 (诱导剂结合与未结合状态的比例)，固定值
    P: RNA聚合酶浓度
    delta_epsilon_pd: RNAP-DNA结合自由能变化
    Imax: 最大表达水平
    beta: 1/kT
    
    返回:
    表达水平
    """
    # 计算二聚体浓度 (假设与TF和DBD浓度有关)
    # 使用质量作用定律简化模型
    K_dimer = 1.0  # 二聚化常数
    dimer_conc = K_dimer * R * DBD
    
    # 计算结合项
    binding_term = (2 * dimer_conc / N_NS) * np.exp(-beta * delta_epsilon_rd)
    
    # 计算相互作用项 (二聚体协同性)
    interaction_term = np.exp(-beta * delta_epsilon_int)
    
    # 计算变构项 (诱导剂效应，固定值)
    allosteric_term = (1 + K)**2 / (1 + K * np.exp(-beta * delta_epsilon_int))**2
    
    # 计算RNAP结合项
    polymerase_term = (P / N_NS) * np.exp(-beta * delta_epsilon_pd)
    
    # 计算启动子占据概率
    p_bound = polymerase_term / (1 + polymerase_term + binding_term * interaction_term * allosteric_term)
    
    # 计算表达水平
    expression = Imax * p_bound
    
    return expression

# 创建浓度网格
R = np.logspace(-3, 2, 200)  # TF浓度范围
DBD = np.logspace(-3, 2, 200)  # DBD浓度范围
R_grid, DBD_grid = np.meshgrid(R, DBD)

# 参数设置 (基于Rob Phillips模型和您提供的参数)
N_NS = 4.6e6  # 大肠杆菌中的非特异性结合位点数
delta_epsilon_rd = -12.0  # TF-DNA结合自由能变化 (kcal/mol)
delta_epsilon_int = -5.0  # 二聚体相互作用自由能变化 (kcal/mol)
delta_epsilon_pd = -8.0  # RNAP-DNA结合自由能变化 (kcal/mol)
K = 100.0  # 变构常数 (固定值，反映诱导剂浓度固定)
P = 1000.0  # RNA聚合酶浓度
Imax = 35.84931505  # 最大表达水平 (使用您提供的值)
beta = 1/0.6  # 1/kT，在室温下约为1/0.6 kcal/mol

# 计算表达水平
expression = rob_phillips_allosteric_dimer_fixed_inducer(R_grid, DBD_grid, N_NS, delta_epsilon_rd, 
                                                       delta_epsilon_int, K, P, delta_epsilon_pd, Imax, beta)

# 计算无TF时的表达水平 (参考状态)
expression_no_tf = rob_phillips_allosteric_dimer_fixed_inducer(0, 0, N_NS, delta_epsilon_rd, 
                                                              delta_epsilon_int, K, P, delta_epsilon_pd, Imax, beta)

# 计算fold change
fold_change = expression / expression_no_tf

# 创建图形
fig, ax = plt.subplots(figsize=(8, 6))

# 使用对数标准化
norm = colors.LogNorm(vmin=fold_change.min(), vmax=fold_change.max())

# 创建热图 - 使用turbo colormap
cp = ax.pcolormesh(R_grid, DBD_grid, fold_change, cmap='turbo', shading='auto', norm=norm)

# 添加颜色条
cbar = fig.colorbar(cp)
cbar.ax.set_ylabel('Fold Change', fontsize=12)

# 添加等高线
contour_levels = [1, 2, 5, 10, 20, 50, 100]
cs = ax.contour(R_grid, DBD_grid, fold_change, levels=contour_levels, colors='white', linewidths=1)
ax.clabel(cs, fmt='%1.0f', fontsize=8)

# 设置坐标轴
ax.set_xlabel('Transcription Factor Concentration (TF)', fontsize=12)
ax.set_ylabel('DBD Concentration', fontsize=12)
ax.set_title('Rob Phillips Allosteric Dimer Model\nFixed Inducer Concentration', fontsize=14)

# 使用对数坐标
ax.set_xscale('log')
ax.set_yscale('log')

# 美化外框
ax.spines['bottom'].set_linewidth(0.5)
ax.spines['left'].set_linewidth(0.5)
ax.spines['top'].set_linewidth(0.5)
ax.spines['right'].set_linewidth(0.5)
plt.tick_params(axis='both', width=0.5, labelsize=10)

# 添加注释说明模型特性
ax.text(0.02, 0.98, 'Fixed Inducer Concentration\n• DBD concentration is continuous\n• Inducer effect captured by allosteric constant K\n• Cooperative TF binding', 
        transform=ax.transAxes, fontsize=10, verticalalignment='top',
        bbox=dict(boxstyle="round,pad=0.3", facecolor="white", alpha=0.8))

plt.tight_layout()

# 保存图像
plt.savefig('rob_phillips_fixed_inducer.pdf', format='pdf', dpi=300, bbox_inches='tight')

# 显示图像
plt.show()

# 额外创建一个分析图，显示在不同DBD浓度下fold change随TF浓度的变化
plt.figure(figsize=(10, 6))

# 选择几个固定的DBD浓度值
fixed_dbd_values = [0.01, 0.1, 1.0, 10.0]
colors_list = ['red', 'blue', 'green', 'purple']

for dbd_val, color in zip(fixed_dbd_values, colors_list):
    # 计算在不同TF浓度下的fold change
    expression_with_tf = rob_phillips_allosteric_dimer_fixed_inducer(R, dbd_val, N_NS, delta_epsilon_rd, 
                                                                   delta_epsilon_int, K, P, delta_epsilon_pd, Imax, beta)
    expression_without_tf = rob_phillips_allosteric_dimer_fixed_inducer(0, 0, N_NS, delta_epsilon_rd, 
                                                                      delta_epsilon_int, K, P, delta_epsilon_pd, Imax, beta)
    fc = expression_with_tf / expression_without_tf
    plt.semilogx(R, fc, color=color, linewidth=2, label=f'DBD = {dbd_val}')

plt.xlabel('Transcription Factor Concentration (TF)', fontsize=12)
plt.ylabel('Fold Change', fontsize=12)
plt.title('Rob Phillips Model: Fold Change vs TF Concentration\nAt Fixed DBD Concentrations (Inducer Fixed)', fontsize=14)
plt.legend()
plt.grid(True, which="both", ls="--", alpha=0.5)

plt.tight_layout()
plt.savefig('rob_phillips_fixed_inducer_cross_section.pdf', format='pdf', dpi=300, bbox_inches='tight')
plt.show()

# 创建一个3D曲面图来更好地展示关系

fig = plt.figure(figsize=(12, 8))
ax = fig.add_subplot(111, projection='3d')

# 创建表面图
surf = ax.plot_surface(np.log10(R_grid), np.log10(DBD_grid), np.log10(fold_change), 
                       cmap='turbo', edgecolor='none', alpha=0.8)

# 设置标签
ax.set_xlabel('log10(TF Concentration)', fontsize=12)
ax.set_ylabel('log10(DBD Concentration)', fontsize=12)
ax.set_zlabel('log10(Fold Change)', fontsize=12)
ax.set_title('3D Surface: Rob Phillips Allosteric Dimer Model\n(Fixed Inducer Concentration)', fontsize=14)

# 添加颜色条
fig.colorbar(surf, ax=ax, shrink=0.5, aspect=20, label='log10(Fold Change)')

plt.tight_layout()
plt.savefig('rob_phillips_3d_surface.pdf', format='pdf', dpi=300, bbox_inches='tight')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.colors as colors

def simple_repression_partition_function(R, DBD, N_NS, delta_epsilon_pd, delta_epsilon_rd, P, beta=1.0/0.6):
    """
    基于Rob Phillips简单抑制模型的配分函数计算
    
    参数:
    R: 转录因子浓度
    DBD: DBD浓度
    N_NS: 非特异性结合位点数量
    delta_epsilon_pd: RNAP-DNA结合自由能变化
    delta_epsilon_rd: TF-DNA结合自由能变化
    P: RNA聚合酶浓度
    beta: 1/kT
    
    返回:
    配分函数Z
    """
    # 计算各项权重
    empty_promoter = 1.0
    rnap_bound = (P / N_NS) * np.exp(-beta * delta_epsilon_pd)
    
    # 假设DBD浓度影响TF的有效浓度
    effective_R = R * DBD / (1 + DBD)  # 简化模型
    
    tf_bound = (2 * effective_R / N_NS) * np.exp(-beta * delta_epsilon_rd)
    
    # 配分函数
    Z = empty_promoter + rnap_bound + tf_bound
    
    return Z

def simple_repression_pbound(R, DBD, N_NS, delta_epsilon_pd, delta_epsilon_rd, P, beta=1.0/0.6):
    """
    计算RNA聚合酶结合概率
    
    参数:
    同simple_repression_partition_function
    
    返回:
    RNA聚合酶结合概率
    """
    Z = simple_repression_partition_function(R, DBD, N_NS, delta_epsilon_pd, delta_epsilon_rd, P, beta)
    
    # RNA聚合酶结合项的权重
    rnap_bound = (P / N_NS) * np.exp(-beta * delta_epsilon_pd)
    
    # RNA聚合酶结合概率
    p_bound = rnap_bound / Z
    
    return p_bound

def simple_repression_foldchange(R, DBD, N_NS, delta_epsilon_pd, delta_epsilon_rd, P, Imax, beta=1.0/0.6):
    """
    计算fold change
    
    参数:
    Imax: 最大表达水平
    其他参数同simple_repression_partition_function
    
    返回:
    fold change
    """
    # 有TF时的表达水平
    expression_with_tf = Imax * simple_repression_pbound(R, DBD, N_NS, delta_epsilon_pd, delta_epsilon_rd, P, beta)
    
    # 无TF时的表达水平 (参考状态)
    expression_without_tf = Imax * simple_repression_pbound(0, DBD, N_NS, delta_epsilon_pd, delta_epsilon_rd, P, beta)
    
    # fold change
    fold_change = expression_with_tf /expression_without_tf
    
    return fold_change

# 创建浓度网格
R = np.logspace(-3, 2, 200)  # TF浓度范围
DBD = np.logspace(-3, 2, 200)  # DBD浓度范围
R_grid, DBD_grid = np.meshgrid(R, DBD)

# 参数设置 (基于Rob Phillips模型和典型值)
N_NS = 4.6e6  # 大肠杆菌中的非特异性结合位点数
delta_epsilon_pd = -8.0  # RNAP-DNA结合自由能变化 (kcal/mol)
delta_epsilon_rd = -15.0  # TF-DNA结合自由能变化 (kcal/mol)
P = 100.0  # RNA聚合酶浓度
Imax = 35.0  # 最大表达水平
beta = 1/0.6  # 1/kT，在室温下约为1/0.6 kcal/mol

# 计算fold change
fold_change = simple_repression_foldchange(R_grid, DBD_grid, N_NS, delta_epsilon_pd, 
                                         delta_epsilon_rd, P, Imax, beta)

# 创建图形
fig, ax = plt.subplots(figsize=(7, 6))

# 使用对数标准化
norm = colors.LogNorm(vmin=fold_change.min(), vmax=fold_change.max())

# 创建热图 - 使用turbo colormap
cp = ax.pcolormesh(R_grid, DBD_grid, fold_change, cmap='coolwarm', shading='auto', norm=norm)

# 添加颜色条
cbar = fig.colorbar(cp)
cbar.ax.set_ylabel('Fold Change', fontsize=12)

# 添加等高线
contour_levels = [0.1, 0.5, 1, 2, 5, 10]
cs = ax.contour(R_grid, DBD_grid, fold_change, levels=contour_levels, colors='white', linewidths=1)
ax.clabel(cs, fmt='%1.1f', fontsize=8)

# 设置坐标轴
ax.set_xlabel('c(TF)', fontsize=12)
ax.set_ylabel('ΔddG', fontsize=12)
ax.set_title('TypeI', fontsize=14)

# 使用对数坐标
ax.set_xscale('log')
ax.set_yscale('log')

# 美化外框
ax.spines['bottom'].set_linewidth(0.5)
ax.spines['left'].set_linewidth(0.5)
ax.spines['top'].set_linewidth(0.5)
ax.spines['right'].set_linewidth(0.5)
plt.tick_params(axis='both', width=0.5, labelsize=10)
'''
# 添加注释说明模型特性
ax.text(0.02, 0.98, 'Simple Repression Model\n• Z = 1 + (P/N_NS)e^(-βΔε_pd) + (2R/N_NS)e^(-βΔε_rd)\n• DBD affects TF effective concentration', 
        transform=ax.transAxes, fontsize=10, verticalalignment='top',
        bbox=dict(boxstyle="round,pad=0.3", facecolor="white", alpha=0.8))
'''
plt.tight_layout()

# 保存图像
plt.savefig('rob_phillips_simple_repression.pdf', format='pdf', dpi=300, bbox_inches='tight')

# 显示图像
plt.show()

# 额外创建一个分析图，显示在不同DBD浓度下fold change随TF浓度的变化
plt.figure(figsize=(10, 6))

# 选择几个固定的DBD浓度值
fixed_dbd_values = [0.01, 0.1, 1.0, 10.0]
colors_list = ['red', 'blue', 'green', 'purple']

for dbd_val, color in zip(fixed_dbd_values, colors_list):
    # 计算在不同TF浓度下的fold change
    fc = simple_repression_foldchange(R, dbd_val, N_NS, delta_epsilon_pd, 
                                    delta_epsilon_rd, P, Imax, beta)
    plt.semilogx(R, fc, color=color, linewidth=2, label=f'DBD = {dbd_val}')

plt.xlabel('Transcription Factor Concentration (TF)', fontsize=12)
plt.ylabel('Fold Change', fontsize=12)
plt.title('Simple Repression Model: Fold Change vs TF Concentration\nAt Fixed DBD Concentrations', fontsize=14)
plt.legend()
plt.grid(True, which="both", ls="--", alpha=0.5)

plt.tight_layout()
plt.savefig('simple_repression_cross_section.pdf', format='pdf', dpi=300, bbox_inches='tight')
plt.show()

# 创建一个3D曲面图来更好地展示关系

fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')

# 创建表面图
surf = ax.plot_surface(np.log10(R_grid), np.log10(DBD_grid), np.log10(fold_change), 
                       cmap='turbo', edgecolor='none', alpha=0.8)

# 设置标签
ax.set_xlabel('log10(TF Concentration)', fontsize=12)
ax.set_ylabel('log10(DBD Concentration)', fontsize=12)
ax.set_zlabel('log10(Fold Change)', fontsize=12)
ax.set_title('3D Surface: Rob Phillips Simple Repression Model', fontsize=14)

# 添加颜色条
fig.colorbar(surf, ax=ax, shrink=0.5, aspect=20, label='log10(Fold Change)')

plt.tight_layout()
plt.savefig('simple_repression_3d_surface.pdf', format='pdf', dpi=300, bbox_inches='tight')
plt.show()

# 计算并显示配分函数各项的贡献
R_sample = 1.0  # 选择一个TF浓度样本
DBD_sample = 1.0  # 选择一个DBD浓度样本

Z = simple_repression_partition_function(R_sample, DBD_sample, N_NS, delta_epsilon_pd, delta_epsilon_rd, P, beta)
empty_contrib = 1.0 / Z
rnap_contrib = ((P / N_NS) * np.exp(-beta * delta_epsilon_pd)) / Z
effective_R = R_sample * DBD_sample / (1 + DBD_sample)
tf_contrib = ((2 * effective_R / N_NS) * np.exp(-beta * delta_epsilon_rd)) / Z

print(f"Partition function components for R={R_sample}, DBD={DBD_sample}:")
print(f"Empty promoter: {empty_contrib:.4f} ({empty_contrib*100:.1f}%)")
print(f"RNAP bound: {rnap_contrib:.4f} ({rnap_contrib*100:.1f}%)")
print(f"TF bound: {tf_contrib:.4f} ({tf_contrib*100:.1f}%)")
print(f"Total Z: {Z:.4f}")
```
